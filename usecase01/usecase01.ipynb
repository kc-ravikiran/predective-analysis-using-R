**Problem Statement:**

The company needs to import the dataset, identify different data types, perform data conversions, apply filtering conditions, and manipulate text data for better business analysis and reporting.
**Objective:**

Activities to be Performed:

4. Convert variables into suitable formats using as.numeric(), as.character(), and as.factor().
5. Perform arithmetic calculations such as finding average mileage and total engine capacity.

```python
%load_ext rpy2.ipython
```

```python
%%R
# ---------------------------------------
# Problem 1: Data Types and Variables
# Using mtcars Dataset
# ---------------------------------------

# Load built-in dataset
data(mtcars)

# Create a new dataframe with car names
cars <- data.frame(
  Car_Name = rownames(mtcars),
  Mileage = mtcars$mpg,
  Engine_Capacity = mtcars$disp,
  Transmission = factor(mtcars$am,
                        labels = c("Automatic", "Manual"))
)

# Display dataset
print("Car Dataset:")
print(head(cars))

# ---------------------------------------
# 1. Identify Data Types
# ---------------------------------------

cat("\nType of Mileage:", typeof(cars$Mileage), "\n")
cat("Class of Mileage:", class(cars$Mileage), "\n")

cat("\nType of Car_Name:", typeof(cars$Car_Name), "\n")
cat("Class of Car_Name:", class(cars$Car_Name), "\n")

cat("\nType of Transmission:", typeof(cars$Transmission), "\n")
cat("Class of Transmission:", class(cars$Transmission), "\n")

# Structure of dataset
cat("\nStructure of Dataset:\n")
str(cars)

# ---------------------------------------
# 2. Data Type Conversions
# ---------------------------------------

cars$Mileage_char <- as.character(cars$Mileage)
cars$Mileage_num <- as.numeric(cars$Mileage_char)

cars$Transmission_char <- as.character(cars$Transmission)
cars$Transmission_factor <- as.factor(cars$Transmission_char)

cat("\nData Type Conversions Completed.\n")

# ---------------------------------------
# 3. Arithmetic Operations
# ---------------------------------------

avg_mileage <- mean(cars$Mileage)

total_engine_capacity <- sum(cars$Engine_Capacity)

cat("\nAverage Mileage:", avg_mileage)
cat("\nTotal Engine Capacity:", total_engine_capacity)

# ---------------------------------------
# 4. Logical Filtering
# ---------------------------------------

high_mileage_cars <- subset(cars, Mileage > 20)

cat("\n\nCars with Mileage Greater Than 20:\n")
print(high_mileage_cars[, c("Car_Name", "Mileage")])

# ---------------------------------------
# 5. Character Manipulation
# ---------------------------------------

# Combine first two car names
combined_names <- paste(cars$Car_Name[1],
                        cars$Car_Name[2],
                        sep = " & ")

cat("\nCombined Names:", combined_names)

# Extract first three letters
first_three_letters <- substr(cars$Car_Name, 1, 3)

cat("\nFirst Three Letters of Car Names:\n")
print(head(first_three_letters))

# Convert car names to uppercase
uppercase_names <- toupper(cars$Car_Name)

cat("\nCar Names in Uppercase:\n")
print(head(uppercase_names))
```
